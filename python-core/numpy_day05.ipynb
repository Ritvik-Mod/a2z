```python
import time
start_time_py = time.perf_counter()
sum = 0
for i in range(1,10**6):
    sum+=i**2
end_time_py = time.perf_counter()

pytime = end_time_py-start_time_py

start_time_vec = time.perf_counter()
x = np.arange(1,10**6)
x = x**2
end_time_vec = time.perf_counter()

vectime = end_time_vec-start_time_vec

print(f"Python time: {pytime}")
print(f"Vec time: {vectime}")
print(f"Pytime/Vectime : {pytime/vectime}")
```
