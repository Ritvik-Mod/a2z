```python
#Q13

def chunk(lst, n):
    return [lst[i:i+n] for i in range(0,len(lst),n)]
print(chunk([1,2,3,4,5],2))
```

```python
#Q14
matrix = [[1,2,3],[4,5,6],[7,8,9]]

print([a for row in matrix for a in row])

flat = [a for row in matrix for a in row]

transpose = list(map(list,zip(*[flat[i:i+3] for i in range(0,len(flat),3)])))
print(transpose)

print(list(map(list,zip(*[[f"{s:4d}" for row in matrix for s in row][i:i+3] for i in range(0,len([f"{s:4d}" for row in matrix for s in row]),3)]))))
```

```python
#Q16
inp = "HelloWorldFooBar"
def to_snake_case(word):
    op = "".join([("_" + c.lower()) if c.isupper() else c for c in word])
    if(op[0]=="_"):
        return op[1:]
    return op

snake = to_snake_case(inp)
print(snake)
```

```python
#Q17
people = [{"name":"Ritvik","score":88.456},{"name":"Ana","score":92.1},{"name":"Bo","score":75.0}]
lst = sorted(people, key = lambda kv : (-kv["score"]))
print(lst)

print("\n".join(f"{d['name']:<10}{d['score']:>10.1f}%" for d in lst))
```

```python
#Q19
def shift(s,k):
    return "".join(
        chr((ord(c)-ord('a')+k)%26 + ord('a')) if c.isupper() 
        else chr((ord(c)-ord('A')+k)%26 + ord('A')) if c.islower() 
        else c for c in s
        )

s = "Hey 1 bro!"
print(shift(s,3))
print(shift(shift(s,3),-3)==s)
```

```python
#Q20
log = "2026-08-10 09:15:33 ERROR disk full on /dev/sda1"
logs = ["2026-08-10 09:15:33 ERROR disk full on /dev/sda1",
        "2026-08-10 08:29:17 INFO disk full on /dev/sda1",
        "2026-08-10 10:47:23 ERROR disk full on /dev/sda1",
        "2026-08-10 09:45:57 WARN disk full on /dev/sda1"]
datestr = log[0:10]
time = log[11:20]
errpart = log[20:26]
msg = log[26:]
datelst = datestr.split("-")
month = {"01":"January", "02":"February",
         "03":"March", "04":"April",
         "05":"May", "06":"June",
         "07":"July", "08":"August",
         "09":"September", "10":"October",
         "11":"November", "12":"December"}
monate = month[datelst[1]]

str = f"[{time}] {errpart:<4}|  {msg}   ({datelst[2]} {monate} {datelst[0]})"
print(str)

errors = [(log[11:19], line.split(maxsplit=3)[3]) for line in logs if line.split()[2]=="ERROR"]
print(errors)
```
